```python
import yaml
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# --- PRX / Publication Style Configuration ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'mathtext.fontset': 'cm',       # Computer Modern (TeX-like math)
    'font.size': 12,
    'axes.labelsize': 14,
    'axes.linewidth': 1.2,          # Moderate line width for frame
    'xtick.direction': 'in',        # Ticks point inwards
    'ytick.direction': 'in',
    'xtick.top': True,              # Ticks on all sides
    'ytick.right': True,
    'xtick.major.size': 6,
    'ytick.major.size': 6,
    'xtick.minor.size': 3,
    'ytick.minor.size': 3,
    'lines.linewidth': 2.0,         # Bold lines for data
    'legend.frameon': False,        # Clean legends
    'figure.dpi': 300,
    'figure.autolayout': False      # We will use constrained_layout
})

def plot_prx_style(yaml_file):
    # --- Load Data ---
    with open(yaml_file, 'r') as f:
        data = yaml.safe_load(f)

    epochs = [d['epoch'] for d in data]
    total_loss = [d['total_loss/train'] for d in data]
    energy_loss = [d.get('loss_component/energy/train') for d in data]
    forces_loss = [d.get('loss_component/forces/train') for d in data]
    stress_loss = [d.get('loss_component/stress/train') for d in data]

    # --- Setup Figure (Double Column Width) ---
    # PRX double column is approx 7.0 inches wide.
    fig, axs = plt.subplots(2, 2, figsize=(7.0, 6.0), sharex=True)
    plt.subplots_adjust(wspace=0.05, hspace=0.05) # Minimize space between plots

    # Define Plot Configurations
    # Format: (Axis, Data, LabelText, Color, PanelTag)
    configs = [
        (axs[0, 0], total_loss, r'Total Loss', 'k', '(a)'),     # k = Black
        (axs[0, 1], energy_loss, r'Energy', '#d62728', '(b)'),  # Red
        (axs[1, 0], forces_loss, r'Forces', '#1f77b4', '(c)'),  # Blue
        (axs[1, 1], stress_loss, r'Stress', '#2ca02c', '(d)'),  # Green
    ]

    for ax, y_data, label_text, color, tag in configs:
        # 1. Plot Data
        ax.plot(epochs, y_data, 'o', color=color, markersize=1, markeredgewidth=0, alpha=0.6)
        
        # 2. Set Log Scale (No label text about it)
        ax.set_yscale('log')

        # 3. Add Internal "Title" (Top Right usually best for loss curves)
        # We use transform=ax.transAxes to position relative to the box (0-1)
        ax.text(0.95, 0.90, label_text, transform=ax.transAxes, 
                fontsize=14, ha='right', va='top', fontfamily='serif')

        # 4. Add Panel Tag (Top Left)
        ax.text(0.05, 0.90, tag, transform=ax.transAxes, 
                fontsize=14, fontweight='bold', ha='left', va='top')

        # 5. Fine-tune Ticks
        # Log locator ensures ticks show up even if range is small
        ax.yaxis.set_major_locator(ticker.LogLocator(base=10.0, numticks=5))
        ax.yaxis.set_minor_locator(ticker.LogLocator(base=10.0, subs='auto', numticks=5))

    # --- Global Labels ---
    # Only put labels on the outer edges to reduce clutter
    
    # Y-Labels (Left Column only)
    axs[0, 0].set_ylabel('Loss', fontsize=14)
    axs[1, 0].set_ylabel('Loss', fontsize=14)

    # X-Labels (Bottom Row only)
    axs[1, 0].set_xlabel('Epoch', fontsize=14)
    axs[1, 1].set_xlabel('Epoch', fontsize=14)

    # --- Formatting cleanup ---
    # Ensure ticks align nicely and remove overlap
    plt.tight_layout()
    
    # Save
    out_file = 'finetuning_loss_results.png'
    plt.savefig(out_file, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"plot saved as '{out_file}'")

if __name__ == "__main__":
    try:
        plot_prx_style('train_metrics.yaml')
    except FileNotFoundError:
        print("Error: 'train_metrics.yaml' not found.")
```

```python
import yaml
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from pathlib import Path

# ----------------------------
# Panel colors (nice, distinct)
# ----------------------------
PANEL_COLORS = {
    "total":  "#4c4c4c",  # dark gray
    "energy": "#D1495B",  # warm red
    "forces": "#3C7DC4",  # calm blue
    "stress": "#2A9D8F",  # teal
}

# Split styles (keep readable, MACE-like clarity)
SPLIT_STYLES = {
    "train": dict(linestyle="-",  alpha=0.95),
    "valid": dict(linestyle="--", alpha=0.90),
    "test":  dict(linestyle=":",  alpha=0.90),
}

# ----------------------------
# Clean plotting style
# ----------------------------
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 13,
    "axes.linewidth": 1.0,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "legend.frameon": False,
    "savefig.dpi": 600,
})

def _load_yaml(yaml_file: str):
    p = Path(yaml_file)
    if not p.exists():
        raise FileNotFoundError(f"'{yaml_file}' not found.")
    with open(p, "r") as f:
        data = yaml.safe_load(f)
    if not isinstance(data, list) or not data:
        raise ValueError("Expected YAML to be a non-empty list of dicts.")
    return data

def _series(data, key):
    return np.asarray([d.get(key, np.nan) for d in data], dtype=float)

def _plot_metric(ax, epochs, data, base_key, title, panel_tag, color, x_max=100, logy=True):
    # detect splits available
    splits_present = []
    for split in ("train", "valid", "test"):
        k = f"{base_key}/{split}"
        if any(k in d for d in data):
            splits_present.append(split)

    if not splits_present:
        ax.text(0.5, 0.5, "metric not found", ha="center", va="center", transform=ax.transAxes)
        return

    # plot each split using SAME color (panel identity) but different line style (split identity)
    for split in splits_present:
        key = f"{base_key}/{split}"
        y = _series(data, key)

        m = np.isfinite(epochs) & np.isfinite(y)
        m &= (epochs <= x_max)
        if logy:
            m &= (y > 0)

        x = epochs[m]
        yy = y[m]

        style = SPLIT_STYLES[split]
        ax.plot(x, yy, color=color, lw=2.2, **style, label=split)
        # subtle points (rasterized keeps PDF light)
        ax.scatter(x, yy, s=10, color=color, alpha=0.18, linewidths=0, rasterized=True)

    if logy:
        ax.set_yscale("log")
        ax.yaxis.set_major_locator(mticker.LogLocator(base=10, numticks=6))
        ax.yaxis.set_minor_locator(mticker.LogLocator(base=10, subs=np.arange(2, 10) * 0.1, numticks=12))
        ax.yaxis.set_major_formatter(mticker.LogFormatterMathtext(base=10))

    # subtle grid
    ax.grid(True, which="major", linewidth=0.6, alpha=0.18)
    ax.grid(True, which="minor", linewidth=0.4, alpha=0.08)

    # inside annotations
    #ax.text(0.02, 0.96, panel_tag, transform=ax.transAxes, ha="left", va="top", fontsize=13, fontweight="bold")
    ax.text(0.98, 0.96, title, transform=ax.transAxes, ha="right", va="top",fontsize=13)

    # x-limits exactly to 0..x_max
    ax.set_xlim(0, x_max)

def plot_mace_like(yaml_file="train_metrics.yaml", out_prefix="finetuning_metrics_mace_like", x_max=100):
    data = _load_yaml(yaml_file)
    epochs = np.asarray([d.get("epoch", np.nan) for d in data], dtype=float)

    fig, axs = plt.subplots(2, 2, figsize=(7.2, 5.6), sharex=True, constrained_layout=True)

    _plot_metric(axs[0, 0], epochs, data, "total_loss",            "Total loss", "(a)", PANEL_COLORS["total"],  x_max=x_max)
    _plot_metric(axs[0, 1], epochs, data, "loss_component/energy", "Energy",     "(b)", PANEL_COLORS["energy"], x_max=x_max)
    _plot_metric(axs[1, 0], epochs, data, "loss_component/forces", "Forces",     "(c)", PANEL_COLORS["forces"], x_max=x_max)
    _plot_metric(axs[1, 1], epochs, data, "loss_component/stress", "Stress",     "(d)", PANEL_COLORS["stress"], x_max=x_max)

    # Remove "Metric"/"Loss" wording completely (no y-labels)
    for ax in axs[:, 0]:
        ax.set_ylabel("")
    for ax in axs[1, :]:
        ax.set_xlabel("Epoch")

    # Nice x ticks up to 100
    axs[1, 0].xaxis.set_major_locator(mticker.MultipleLocator(20))
    axs[1, 0].xaxis.set_minor_locator(mticker.MultipleLocator(10))

    # Single legend (train/valid/test) using the first axis
    handles, labels = axs[0, 0].get_legend_handles_labels()
    if handles:
        fig.legend(handles, labels, loc="upper center", ncol=3, bbox_to_anchor=(0.5, 1.02))

    png = f"{out_prefix}.png"
    pdf = f"{out_prefix}.pdf"
    fig.savefig(png, bbox_inches="tight")
    fig.savefig(pdf, bbox_inches="tight")
    plt.show()
    print(f"Saved: {png} and {pdf}")

if __name__ == "__main__":
    plot_mace_like("train_metrics.yaml", out_prefix="finetuning_metrics_mace_like", x_max=100)
```

```python
import yaml
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from pathlib import Path

# ----------------------------
# Panel colors
# ----------------------------
PANEL_COLORS = {
    "total":  "#4c4c4c",
    "energy": "#D1495B",
    "forces": "#3C7DC4",
    "stress": "#2A9D8F",
}

SPLIT_STYLES = {
    "train": dict(linestyle="-",  alpha=0.95),
    "valid": dict(linestyle="--", alpha=0.90),
    "test":  dict(linestyle=":",  alpha=0.90),
}

# ----------------------------
# Style
# ----------------------------
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 12,
    "axes.linewidth": 1.0,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
    "legend.frameon": False,
    "savefig.dpi": 600,
    "pdf.fonttype": 42,
})

def _load_yaml(yaml_file):
    p = Path(yaml_file)
    if not p.exists():
        raise FileNotFoundError(f"{yaml_file} not found")
    with open(p) as f:
        data = yaml.safe_load(f)
    return data

def _series(data, key):
    return np.asarray([d.get(key, np.nan) for d in data], float)

def _plot_metric(ax, epochs, data, base_key, title, color, x_max=100):
    splits = [s for s in ("train", "valid", "test")
              if any(f"{base_key}/{s}" in d for d in data)]

    for split in splits:
        y = _series(data, f"{base_key}/{split}")
        m = np.isfinite(epochs) & np.isfinite(y) & (epochs <= x_max) & (y > 0)

        ax.plot(
            epochs[m], y[m],
            color=color,
            lw=2.2,
            label=split,
            **SPLIT_STYLES[split]
        )
        ax.scatter(
            epochs[m], y[m],
            s=10, color=color, alpha=0.18,
            linewidths=0, rasterized=True
        )

    ax.set_yscale("log")
    ax.set_xlim(0, x_max)

    ax.yaxis.set_major_locator(mticker.LogLocator(base=10))
    ax.yaxis.set_minor_locator(mticker.LogLocator(base=10, subs=np.arange(2, 10) * 0.1))
    ax.yaxis.set_major_formatter(mticker.LogFormatterMathtext())

    ax.grid(True, which="major", linewidth=0.6, alpha=0.18)
    ax.grid(True, which="minor", linewidth=0.4, alpha=0.08)

    # Title safely inside panel (no line above it)
    ax.text(
        0.5, 0.91, title,
        transform=ax.transAxes,
        ha="right", va="top",
        fontsize=13
    )

    ax.tick_params(axis="both", which="major", pad=4)

    return splits  # return splits so we know later whether to show legend

def plot_mace_like(yaml_file="train_metrics.yaml", x_max=100):
    data = _load_yaml(yaml_file)
    epochs = np.asarray([d["epoch"] for d in data], float)

    fig, axs = plt.subplots(
        2, 2,
        figsize=(7.2, 5.6),
        sharex=True,
        constrained_layout=True
    )

    fig.set_constrained_layout_pads(
        w_pad=0.02, h_pad=0.04,
        wspace=0.08, hspace=0.10
    )

    splits_all = []
    splits_all += _plot_metric(axs[0, 0], epochs, data, "total_loss",            "Total loss", PANEL_COLORS["total"],  x_max)
    splits_all += _plot_metric(axs[0, 1], epochs, data, "loss_component/energy", "Energy",     PANEL_COLORS["energy"], x_max)
    splits_all += _plot_metric(axs[1, 0], epochs, data, "loss_component/forces", "Forces",     PANEL_COLORS["forces"], x_max)
    splits_all += _plot_metric(axs[1, 1], epochs, data, "loss_component/stress", "Stress",     PANEL_COLORS["stress"], x_max)

    # X labels only bottom row
    for ax in axs[1, :]:
        ax.set_xlabel("Epoch")

    # Remove y labels completely
    for ax in axs[:, 0]:
        ax.set_ylabel("")

    # X ticks
    for ax in axs[1, :]:
        ax.xaxis.set_major_locator(mticker.MultipleLocator(20))
        ax.xaxis.set_minor_locator(mticker.MultipleLocator(10))

    # ✅ ONLY show legend if more than one split exists
    unique_splits = sorted(set(splits_all))
    if len(unique_splits) > 1:
        handles, labels = axs[0, 0].get_legend_handles_labels()
        fig.legend(
            handles, labels,
            loc="upper center",
            ncol=len(unique_splits),
            bbox_to_anchor=(0.5, 1.05),
            handlelength=2.0,
            columnspacing=1.5
        )

    # Ensure no axis-level legends remain
    for ax in axs.flat:
        if ax.legend_:
            ax.legend_.remove()

    fig.savefig("finetuning_metrics.pdf", bbox_inches="tight")
    fig.savefig("finetuning_metrics.png", bbox_inches="tight")
    plt.show()

if __name__ == "__main__":
    plot_mace_like()
```
